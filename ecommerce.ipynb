```python
# Replace ['N/A', "NULL", " "]   with  NaN

df.replace(['N/A', "NULL", " "], np.nan, inplace=True)
```

```python
# Remove Extra Spaces 

df = df.apply(lambda x : x.str.strip() if x.dtype == 'str' else x)
```

```python
df.columns
```

```python
df.head()
```

```python
# Standardize Text Fromat 
# Convert Text into proper Case. 

df['Customer_Name'] = df['Customer_Name'].str.title()
df['City'] = df['City'].str.title()
df['State'] = df['State'].str.title()
df['Product'] = df['Product'].str.title()
df['Category'] = df['Category'].str.title()
```

```python
df.head()
```

```python
# Valiate Email Id 
# Keep only valid email record.

df = df[df['Email'].str.contains('@', na=False)]
```

```python
# Convert Date columns 

df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')   # NAT
df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'], errors='coerce')   # NAT
```

```python
df.columns
```

```python
# Convert Numeric Columns 

cols = ['Qty', 'Unit_Price', 'Discount']

for i in cols: 
    df[i] = pd.to_numeric(df[i], errors='coerce')
```

```python
# Remove Invalid Quantity 

df = df[df['Qty'] > 0 ]
```

```python
# Fill Missing Values 

df['Discount'] = df['Discount'].fillna(0)
df['Phone'] = df['Phone'].fillna('Unknown')
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
df['Unit_Price'] = df['Unit_Price'].fillna(
    df.groupby('Product')['Unit_Price'].transform('median')
)
```

```python
df['Unit_Price'] = df['Unit_Price'].fillna(
    df.groupby('Product')['Unit_Price'].transform('median')
)
```

```python
# Feature Engineering 

# Total Sales 


df['Sales'] = df['Qty'] * df['Unit_Price']
```

```python
df.head()
```

```python
# Net Amount 

df["Net_Amount"] = df["Sales"] - (df['Sales'] * df["Discount"] / 100)
```

```python
df.head()
```

```python
# Profit (Ex = 20%)

df["Profit"] = df['Net_Amount'] * 0.20
```

```python
df.head()
```

```python
# Order_Month

df["Order_Month"] = df['Order_Date'].dt.month_name()
```

```python
df
```

```python
# Order_Year

df["Order_Year"] = df['Order_Date'].dt.year

```

```python
df['Weekday'] = df['Order_Date'].dt.day_name()
```
